```python
import sys
sys.path.append('../')

import pandas as pd
import matplotlib.pyplot as plt
from src.preprocess import load_and_merge_training_data
```

```python
training_session = ['FP1', 'FP2', 'FP3', 'Q']
df = load_and_merge_training_data(training_session)
print(f'Training Baseline Matrix Shape: {df.shape}')
```

```python
fig, ax = plt.subplots(figsize=(12,5))
fig.set_facecolor(color='#15151E')
ax.set_facecolor(color='#15151E')

ax.plot(df['Speed'].iloc[:1000], color='#FF1E00', label='Speed (km/h)')

ax.set_title('Sample Telemetry Trace: Clean Qualifying Laps', weight='bold', color='#F5F5F5')

ax.set_xlabel('Time Step (ms)', color='#F5F5F5')
ax.set_ylabel('Speed (km/h)', color='#F5F5F5')

ax.tick_params(axis='both', colors='#F5F5F5')
ax.legend(facecolor='#15151E', edgecolor='#F5F5F5', labelcolor='#F5F5F5')
```
